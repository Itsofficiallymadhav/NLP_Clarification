```python
import json
from tqdm import tqdm

N = 20
results = []
passed = 0

for i in tqdm(range(N)):
    sample = dataset["train"][i]
    try:
        response = generate(sample["text"], test_example=sample["test_list"][0])
    except Exception as e:
        results.append({"idx": i, "passed": False, "error": f"generation error: {e}"})
        continue

    success, error = evaluate(
        response,
        sample["test_list"],
        setup_code=sample.get("test_setup_code", "")
    )

    if success:
        passed += 1

    results.append({
        "idx": i,
        "task_id": sample.get("task_id", i),
        "passed": success,
        "error": error,
        "code": extract_code(response)
    })

print(f"\nPassed: {passed}/{N} ({passed/N*100:.1f}%)")

with open("../results/baseline_run_20.json", "w") as f:
    json.dump(results, f, indent=2)
```
